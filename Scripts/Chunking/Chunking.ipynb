```python
import nltk
from nltk.chunk import RegexpParser
from nltk.chunk import tree2conlltags

nltk.download('punkt')
```

## Data Preprocessing

```python
# Teks 5 kalimat pertama dari 1_tagged_corpus.txt
teks_5_kalimat = """
Pada/IN hari/NND Kamis/NNP ,/Z 1/CD Oktober/NNP 2026/CD ,/Z Presiden/NNP Prabowo/NNP Subianto/NNP secara/RB resmi/JJ melakukan/VB perombakan/NN atau/CC perombakan/NN kabinet/NN yang/SC ketujuh/OD kalinya/NND di/IN Istana/NNP Merdeka/NNP ./Z
Dalam/IN perombakan/NN jajaran/NN Kabinet/NNP Merah/JJ Putih/JJ tersebut/PR ,/Z sejumlah/CD posisi/NN strategis/JJ mengalami/VB pergantian/NN ,/Z mulai/NN dari/IN tingkatan/NN menteri/NN hingga/IN wakil/NN menteri/NN ./Z
Selain/NN itu/PR ,/Z dilantik/VB pula/RB beberapa/CD pejabat/NN baru/JJ yang/SC diharapkan/VB mampu/MD menambah/VB dimensi/NN kebijakan/NN di/IN tingkat/NN pemerintahan/NN pusat/NN ./Z
Perubahan/NN struktural/NN ini/PR ditandai/VB dengan/IN adanya/NN penambahan/NN posisi/NN wakil/NN menteri/NN ./Z
Secara/RB konseptual/NN ,/Z penambahan/NN ini/PR bertujuan/VB untuk/IN memperluas/VB jangkauan/NN implementasi/NN kebijakan/NN publik/JJ ./Z
"""

# Konversi string ke format List of Tuples: [(Kata, Tag), (Kata, Tag), ...]
kumpulan_kalimat = []
for baris in teks_5_kalimat.strip().split('\n'):
    kalimat_tuple = []
    for token in baris.split():
        if '/' in token:
            kata, tag = token.rsplit('/', 1)
            kalimat_tuple.append((kata, tag))
    kumpulan_kalimat.append(kalimat_tuple)

# Cek hasil konversi kalimat pertama untuk validasi
print("Contoh struktur data kalimat pertama:")
print(kumpulan_kalimat[0])
```

## Grammar Definition

```python
# Pembuatan grammar Noun Phrase (NP) berdasarkan Tagset II UI-POSTAG
np_grammar = """
    NP: {<CD|OD>?<NNP|NN|NND>+<JJ>*<PR>?}
"""

# Menginisialisasi parser menggunakan grammar yang telah dibuat
chunk_parser = RegexpParser(np_grammar)

print("RegexpParser berhasil diinisialisasi dengan grammar NP.")
```

## Chunking Process

```python
# Membuka file untuk menyimpan output secara write mode ("w")
file_tree = open("2_hasil_np_tree.txt", "w", encoding="utf-8")
file_iob = open("2_hasil_np_iob.txt", "w", encoding="utf-8")

print("=== HASIL EKSTRAKSI PARSE TREE DAN IOB ===")

for idx, kalimat in enumerate(kumpulan_kalimat, 1):
    # Eksekusi metode chunking
    tree = chunk_parser.parse(kalimat)

    # 1. Mengambil representasi String Parse Tree
    hasil_tree_str = str(tree)
    file_tree.write(f"Kalimat ke-{idx}:\n{hasil_tree_str}\n\n")
    print(f"\n--- Kalimat ke-{idx} ---")
    print(f"Parse Tree:\n{hasil_tree_str}")

    # 2. Mengambil representasi format IOB
    iob_tags = tree2conlltags(tree)
    file_iob.write(f"Kalimat ke-{idx}:\n")
    for token in iob_tags:
        # Menulis format tabular: Kata (tab) Tag (tab) IOB
        file_iob.write(f"{token[0]}\t{token[1]}\t{token[2]}\n")
    file_iob.write("\n")

# Menutup akses file agar memori aman tersimpan
file_tree.close()
file_iob.close()

print("\n\nPROSES SELESAI!")
print("Berkas '2_hasil_np_tree.txt' dan '2_hasil_np_iob.txt' sukses di-*generate*.")
```
